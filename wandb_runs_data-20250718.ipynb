```python
import wandb
import pandas as pd

api = wandb.Api(timeout=180)

entity = "tlinjordet-university-of-oslo"
project = "OAReactDiff-SCAN"


# Get a list of runs from a project
# You can filter by tags, state, config, etc.
runs = api.runs(f"{entity}/{project}")

all_history_dfs = []

# Iterate through each run and pull its history
for run in runs:
    # Get the run history as a DataFrame
    history_df = run.history(pandas=True)
    
    # Add a column to identify the run
    history_df['run_id'] = run.id
    history_df['run_name'] = run.name

    all_history_dfs.append(history_df)

# Concatenate all the individual DataFrames into a single one
combined_df = pd.concat(all_history_dfs)

print("Combined DataFrame from multiple runs:")
print(combined_df.head())
```

```python
import wandb
import pandas as pd

# 1. Initialize the W&B API
api = wandb.Api(timeout=180)

# Replace these with your actual entity, project, and run ID
#entity = "your_entity_name"  # e.g., "my_username" or "my_team"
entity = "tlinjordet-university-of-oslo"

#project = "your_project_name" # e.g., "my-first-project"
project = "OAReactDiff-SCAN"

#run_id = "your_run_id"       # e.g., "s14s35c3"
#run_id = "9w-cutoff_12-lr5e-4-StepLR-SCAN-leftnet3b7e89f781c1"
run_id = "runs/j0eig6iu" # works
#run_id = "j0eig6iu" # does not work

try:
    run = api.run(f"{entity}/{project}/{run_id}")

    print("Streaming history data...")
    # Stream the history and convert each step into a dictionary
    history_list = [row for row in run.scan_history()]

    # Convert the list of dictionaries into a DataFrame
    history_df = pd.DataFrame(history_list)

    print("Training log data successfully loaded into a DataFrame.")
    print(history_df.head())

except Exception as e:
    print(f"An error occurred: {e}")
```
